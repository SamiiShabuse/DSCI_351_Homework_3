## 1.1: Load Data

```python
import pandas as pandas
```

```python
MOVIE_PATH = '../data/tmdb_5000_movies.csv'
CREDIT_PATH = '../data/tmdb_5000_credits.csv'
```

```python
movies = pandas.read_csv(MOVIE_PATH)
credits = pandas.read_csv(CREDIT_PATH)
```

```python
movies.head(1)
```

```python
credits.head(1)
```

```python
print("Movies shape: ", movies.shape)
print("Credits shape: ", credits.shape)
```

```python
print("Movie Columns: ", movies.columns)
print("Credit Columns: ", credits.columns)
```

```python
merged_movie_df = movies.merge(
    credits,
    left_on='id',
    right_on='movie_id',
    suffixes=('_movie', '_credit')
)
```

```python
print("Merged shape: ", merged_movie_df.shape)
```
