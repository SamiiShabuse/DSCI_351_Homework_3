# Step 1
Load two files, observe the data, and merge them based on the common column(s).
Perform data cleaning on genres feature first. How many unique genres are there in the
dataset? List top 3 genres by count, provide the genre type and count value.

## 1.1: Load Data

```python
import pandas as pandas
import ast
from collections import Counter
```

```python
MOVIE_PATH = '../data/tmdb_5000_movies.csv'
CREDIT_PATH = '../data/tmdb_5000_credits.csv'
```

```python
movies = pandas.read_csv(MOVIE_PATH)
credits = pandas.read_csv(CREDIT_PATH)
```

```python
movies.head(1)
```

```python
credits.head(1)
```

```python
print("Movies shape: ", movies.shape)
print("Credits shape: ", credits.shape)
```

```python
print("Movie Columns: ", movies.columns)
print("Credit Columns: ", credits.columns)
```

```python
merged_movie_df = movies.merge(
    credits,
    left_on='id',
    right_on='movie_id',
    suffixes=('_movie', '_credit')
)
```

```python
merged_movie_df.columns
```

```python
merged_movie_df.head(1)
```

```python
print("Merged shape: ", merged_movie_df.shape)
```

## 1.2: Clean on Genres feature


```python
def clean_genres(genres):
    try:
        values = ast.literal_eval(genres)
        return [v['name'] for v in values]
    except:
        return []

merged_movie_df['genres_cleaned'] = merged_movie_df['genres'].apply(clean_genres)

merged_movie_df[['title_movie', 'genres', 'genres_cleaned']].head()
```

## Step 1.3 Count unique genres and list top 3 genres by count

```python
genre_counts = Counter()
for genres in merged_movie_df['genres_cleaned']:
    genre_counts.update(genres)

print("Unique genres: ", len(genre_counts))
print("Top 3 genres: ", genre_counts.most_common(3))
```

# Step 2
Now look into the cast feature. Clean the column and find out how many unique
casts in the dataset. List top 3 casts by count, provide the cast’s name and their count value.

## Step 2.1: Clean on Cast 

```python
def clean_cast(cast):
    try:
        values = ast.literal_eval(cast)
        return [v['name'] for v in values]
    except:
        return []
    
# Clean the cast column
merged_movie_df["cast_cleaned"] = merged_movie_df["cast"].apply(clean_cast)

# Check the cleaned result
merged_movie_df[["title_movie", "cast", "cast_cleaned"]].head()
```

## 2.2: Count unique casts and list top 3 casts by count

```python
cast_counts = Counter()
for casts in merged_movie_df['cast_cleaned']:
    cast_counts.update(casts)

print("Unique casts: ", len(cast_counts))
print("Top 3 casts: ", cast_counts.most_common(3))
```
