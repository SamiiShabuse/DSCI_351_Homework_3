## 1.2: Clean on Genres feature


```python
def clean_genres(genres):
    try:
        values = ast.literal_eval(genres)
        return [v['name'] for v in values]
    except:
        return []

merged_movie_df['genres_cleaned'] = merged_movie_df['genres'].apply(clean_genres)

merged_movie_df[['title_movie', 'genres', 'genres_cleaned']].head()
```

## Step 1.3 Count unique genres and list top 3 genres by count

```python
genre_counts = Counter()
for genres in merged_movie_df['genres_cleaned']:
    genre_counts.update(genres)

print("Unique genres: ", len(genre_counts))
print("Top 3 genres: ", genre_counts.most_common(3))
```

```python
def clean_cast(cast):
    try:
        values = ast.literal_eval(cast)
        return [v['name'] for v in values]
    except:
        return []
    
# Clean the cast column
merged_movie_df["cast_cleaned"] = merged_movie_df["cast"].apply(clean_cast)

# Check the cleaned result
merged_movie_df[["title_movie", "cast", "cast_cleaned"]].head()
```

## Step 2.2: Count unique casts and list top 3 casts by count

```python
cast_counts = Counter()
for casts in merged_movie_df['cast_cleaned']:
    cast_counts.update(casts)

print("Unique casts: ", len(cast_counts))
print("Top 3 casts: ", cast_counts.most_common(3))
```

## 6: Find Top 8 Neighbors

```python
def get_neighbors(movie_id, k=8):
    distances = []
    for idx, row in df.iterrows():
        if row['id'] == movie_id: continue
        d = calculate_total_distance(movie_id, row['id'])
        if d > 0:
            distances.append((row['id'], d))
    
    # Sort by distance (ascending)
    distances.sort(key=lambda x: x[1])
    return distances[:k]
```

## 7: Predict Rating

```python
def predict_rating(title, k=10):
    # Find movie id based on title
    movie_row = df[df['title'] == title].iloc[0]
    m_id = movie_row['id']
    actual = movie_row['vote_average']
    
    # Get top k neighbors
    neighbors = get_neighbors(m_id, k=k)
    neighbor_ids = [n[0] for n in neighbors]
    
    # Average vote_average of top k movies as predicted rating
    neighbor_ratings = df[(df['id'].isin(neighbor_ids)) & (df['vote_average'] > 0)]['vote_average']
    prediction = neighbor_ratings.mean()
    
    return prediction, actual
```
